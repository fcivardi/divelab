# Lesson 01 – Hydrostatic Pressure

*Every dynamic model begins with its environment.*

## Theory

$$\frac{dP}{dz}=\rho g$$

$$P=P_0+\rho gz$$

```
from dataclasses import dataclass
import numpy as np
import matplotlib.pyplot as plt

@dataclass(frozen=True)
class Environment:
    density: float=1025.0
    gravity: float=9.80665
    atmospheric_pressure: float=101325.0
    def pressure(self,depth):
        return self.atmospheric_pressure+self.density*self.gravity*depth
    def pressure_bar(self,depth):
        return self.pressure(depth)/100000
    def gauge_pressure(self,depth):
        return self.pressure(depth)-self.atmospheric_pressure

sea=Environment()
for d in [0,10,20,30,40]:
    print(d, sea.pressure_bar(d))

depth=np.linspace(0,40,200)
plt.plot(depth,[sea.pressure_bar(d) for d in depth])
plt.grid(True)
plt.xlabel("Depth [m]")
plt.ylabel("Pressure [bar]")
plt.show()
```
